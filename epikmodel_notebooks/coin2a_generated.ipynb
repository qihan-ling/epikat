## Booleans and States

```python
Bool = hfst.regex('["0"|"1"]')
Bool.view()
```

```python
# Unrestricted state
defs = {"Bool": Bool}
St0 = hfst.regex('[Bool Bool Bool Bool]', definitions=defs)
St0.view()
```

```python
H1 = hfst.regex('[1 Bool Bool Bool]', definitions=defs);
T1 = hfst.regex('[Bool 1 Bool Bool]', definitions=defs);
H2 = hfst.regex('[Bool Bool 1 Bool]', definitions=defs);
T2 = hfst.regex('[Bool Bool Bool 1]', definitions=defs);
```

```python
defs = {"Bool": Bool, "St0": St0, "H1": H1, "T1": T1, "H2": H2, "T2": T2}
```

```python
# Apply constraints from assertions
St = St0.copy()
St.intersect(hfst.regex('((((St0 - (H1 & T1)) & (H1 | T1)) & (St0 - (H2 & T2))) & (H2 | T2))', definitions=defs))
St.minimize()
St.view()
```

```python
defs.update({"St": St})
```

```python
# Constrain state variables by St
H1.intersect(St)
H1.minimize()
T1.intersect(St)
T1.minimize()
H2.intersect(St)
H2.minimize()
T2.intersect(St)
T2.minimize()
defs.update({"H1": H1, "T1": T1, "H2": H2, "T2": T2})
```

## Operations

```python
# State complement
def Nst(X):
    nst = St.copy()
    nst.minus(X)
    return nst
```

```python
# State complements
nH1 = Nst(H1)
nT1 = Nst(T1)
nH2 = Nst(H2)
nT2 = Nst(T2)
defs.update({"nH1": nH1, "nT1": nT1, "nH2": nH2, "nT2": nT2})
```

## Ekat Algebra

```python
# Delete the second state in a block of two states
Squash = hfst.regex('St -> 0 || St _', definitions=defs)
defs.update({"Squash": Squash})
```

```python
UnequalStPair = hfst.regex('[H1 nH1] | [nH1 H1] | [T1 nT1] | [nT1 T1] | [H2 nH2] | [nH2 H2] | [T2 nT2] | [nT2 T2]', definitions=defs)
UnequalStPair.minimize()
defs.update({"UnequalStPair": UnequalStPair})
```

```python
Wf0 = hfst.regex('~[$ UnequalStPair]', definitions=defs)
defs.update({"Wf0": Wf0})
```

```python
# Ekat concatenation
def Cn(X, Y):
    Z = X.copy()
    Z.concatenate(Y)
    Z.intersect(Wf0)
    Z.compose(Squash)
    Z.output_project()
    Z.minimize()
    return Z
```

```python
# Kleene plus
def Kpl(X):
    Z = X.copy()
    Z.repeat_plus()
    Z.intersect(Wf0)
    Z.compose(Squash)
    Z.output_project()
    Z.determinize()
    Z.minimize()
    return Z
```

```python
# Kleene star
def Kst(X):
    Z = Kpl(X)
    Z.disjunct(St)
    Z.determinize()
    Z.minimize()
    return Z
```
